**#análise descritiva das variáveis qualitativas**

#calculando frequências

```python
frenquecia_sexos = dataFrame.Sexo.value_counts()
frequencia_relativa_sexos = dataFrame.Sexo.value_counts(normalize=True)
```

```python
sexos = pd.DataFrame(
    {
        "Frequência": frenquecia_sexos,
        "Percentual (%)": frequencia_relativa_sexos * 100
    }
)
sexos
```

#dando nome às linhas

```python
capa_sexos = {
    0: "Masculino",
    1: "Feminimo"
}

sexos.rename(index=capa_sexos, inplace=True)

sexos
```

#criando uma tabela que mostra o cruzamento entre as váriáveis Sexo e Cores (frequência)

```python
capa_cores = {
    0: "Indígena",
    2: "Branca",
    4: "Preta",
    6: "Amarela",
    8: "Parda",
    9: "Sem declaração"
}

sexos_e_cores = pd.crosstab(dataFrame.Sexo, dataFrame.Cor)
sexos_e_cores.rename(index=capa_sexos, columns=capa_cores, inplace=True)
sexos_e_cores
```

```python
sexos_e_cores_percentuais = pd.crosstab(dataFrame.Sexo, dataFrame.Cor, normalize=True) * 100
sexos_e_cores_percentuais.rename(index=capa_sexos, columns=capa_cores, inplace=True)
sexos_e_cores_percentuais
```

#criando uma tabela que mostra o cruzamento entre as váriáveis Sexo e Cores e faz uma média da renda

```python
sexos_e_cores_renda = pd.crosstab(dataFrame.Sexo, dataFrame.Cor, aggfunc='mean', values=dataFrame.Renda)
sexos_e_cores_renda.rename(index=capa_sexos, columns=capa_cores, inplace=True)
sexos_e_cores_renda
```

**#análise descritiva das variáveis quantitativas**

#criando segmentações de renda

```python
agrupamento = [
    0, 1576, 3152, 7880, 15760, 200000
]

rotulos = [
    'E', 'D', 'C', 'B', 'A'
]

dataFrame_classificado_pela_renda = pd.cut(
    x=dataFrame.Renda,
    bins=agrupamento,
    labels=rotulos,
    include_lowest=True
)

dataFrame_classificado_pela_renda

```

#frequencia de cada renda

```python
frequencia_renda = pd.value_counts(
    pd.cut(
        x=dataFrame.Renda,
        bins=agrupamento,
        labels=rotulos,
        include_lowest=True
    )
)

frequencia_relativa_renda = pd.value_counts(
    pd.cut(
        x=dataFrame.Renda,
        bins=agrupamento,
        labels=rotulos,
        include_lowest=True
    ),
    normalize=True
)

df_frequencia_renda = pd.DataFrame(
    {
        "Frequencia": frequencia_renda,
        "Percentual (%)": frequencia_relativa_renda
    }
)

df_frequencia_renda
```
